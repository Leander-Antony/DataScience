```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# Define inputs
x = np.linspace(-10, 10, 400)

# Activation functions
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

# Softmax works on vectors, not scalars
x_softmax = np.array([x, x+2, x-2])  # 3-class example
softmax = tf.nn.softmax(x_softmax, axis=0).numpy()

# Plot
plt.figure(figsize=(12, 8))

# Sigmoid
plt.subplot(2,2,1)
plt.plot(x, sigmoid, 'r')
plt.title("Sigmoid Function")
plt.grid()

# Tanh
plt.subplot(2,2,2)
plt.plot(x, tanh, 'g')
plt.title("Tanh Function")
plt.grid()

# ReLU
plt.subplot(2,2,3)
plt.plot(x, relu, 'b')
plt.title("ReLU Function")
plt.grid()

# Softmax (example with 3 classes)
plt.subplot(2,2,4)
plt.plot(x, softmax[0], label="Class 1")
plt.plot(x, softmax[1], label="Class 2")
plt.title("Softmax Function (2 classes)")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```
